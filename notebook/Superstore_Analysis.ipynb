```python
import pandas as pd
import matplotlib.pyplot as plt

# Load cleaned dataset
df = pd.read_csv("../data/cleaned/Superstore_Cleaned.csv")

# Display first 5 rows
df.head()
```

```python
# Business KPIs
total_sales = df["Sales"].sum()
total_profit = df["Profit"].sum()
total_orders = df["Order ID"].nunique()
average_sales = df["Sales"].mean()

print("===== BUSINESS KPIs =====")
print(f"Total Sales   : {total_sales:,.2f}")
print(f"Total Profit  : {total_profit:,.2f}")
print(f"Total Orders  : {total_orders}")
print(f"Average Sales : {average_sales:,.2f}")
```

```python
category_sales = df.groupby("Category")["Sales"].sum()

print(category_sales)
```

```python
category_sales.plot(
    kind="bar",
    figsize=(8,5),
    title="Sales by Category"
)

plt.xlabel("Category")
plt.ylabel("Sales")
plt.xticks(rotation=0)
plt.show()
```

```python
category_profit = df.groupby("Category")["Profit"].sum()

category_profit.plot(
    kind="bar",
    figsize=(8,5),
    title="Profit by Category"
)

plt.xlabel("Category")
plt.ylabel("Profit")
plt.xticks(rotation=0)
plt.show()
```

```python
region_sales = df.groupby("Region")["Sales"].sum()

region_sales.plot(
    kind="bar",
    figsize=(8,5),
    title="Sales by Region"
)

plt.xlabel("Region")
plt.ylabel("Sales")
plt.xticks(rotation=0)
plt.show()
```

```python
segment_sales = df.groupby("Segment")["Sales"].sum()

segment_sales.plot(
    kind="pie",
    autopct="%1.1f%%",
    figsize=(6,6),
    ylabel="",
    title="Sales by Segment"
)

plt.show()
```

```python
plt.savefig("../images/category_sales.png", dpi=300, bbox_inches="tight")
```

```python
top_products = df.groupby("Product Name")["Sales"].sum().sort_values(ascending=False).head(10)

print(top_products)
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

# Load cleaned dataset
df = pd.read_csv("../data/cleaned/Superstore_Cleaned.csv")

# Convert Order Date to datetime
df["Order Date"] = pd.to_datetime(df["Order Date"])
```

```python
state_sales = df.groupby("State")["Sales"].sum().sort_values(ascending=False)

print(state_sales.head(10))
```

```python
plt.figure(figsize=(12,6))

state_sales.head(10).plot(kind="bar")

plt.title("Top 10 States by Sales")
plt.xlabel("State")
plt.ylabel("Sales")

plt.xticks(rotation=45)

plt.tight_layout()

plt.savefig("../images/top10_states_sales.png", dpi=300)

plt.show()
```

```python
subcategory_sales = df.groupby("Sub-Category")["Sales"].sum().sort_values(ascending=False)

print(subcategory_sales)
```

```python
plt.figure(figsize=(12,6))

subcategory_sales.plot(kind="bar")

plt.title("Sales by Sub-Category")
plt.xlabel("Sub-Category")
plt.ylabel("Sales")

plt.xticks(rotation=45)

plt.tight_layout()

plt.savefig("../images/subcategory_sales.png", dpi=300)

plt.show()
```

```python
top_profit_products = df.groupby("Product Name")["Profit"].sum().sort_values(ascending=False).head(10)

print(top_profit_products)
```

```python
plt.figure(figsize=(12,6))

top_profit_products.plot(kind="bar")

plt.title("Top 10 Profitable Products")
plt.xlabel("Product Name")
plt.ylabel("Profit")

plt.xticks(rotation=45, ha="right")

plt.tight_layout()

plt.savefig("../images/top10_profitable_products.png", dpi=300)

plt.show()
```

```python
correlation = df[["Sales", "Profit", "Quantity", "Discount"]].corr()

print(correlation)
```

```python
plt.figure(figsize=(6,5))

plt.imshow(correlation, cmap="coolwarm")

plt.colorbar()

plt.xticks(range(len(correlation.columns)), correlation.columns, rotation=45)
plt.yticks(range(len(correlation.columns)), correlation.columns)

plt.title("Correlation Matrix")

plt.tight_layout()

plt.savefig("../images/correlation_matrix.png", dpi=300)

plt.show()
```
